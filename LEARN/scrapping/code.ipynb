```python
import pandas as pd
import numpy as np
import os

# 1. Définition du chemin d'entrée
# Si le fichier est dans le même dossier que ce script, utilisez juste le nom
path = "jobs_wuzzuf_tableau.csv"

if not os.path.exists(path):
    print(f"ERREUR : Le fichier '{path}' est introuvable dans {os.getcwd()}")
else:
    # Lecture avec encodage utf-8 pour éviter les erreurs de caractères
    df = pd.read_csv(path, encoding='utf-8')

    # --- NETTOYAGE ---

    # Supprimer les doublons basés sur le lien (plus précis que toute la ligne)
    df.drop_duplicates(subset=['Lien'], inplace=True) if 'Lien' in df.columns else df.drop_duplicates(inplace=True)

    # Supprimer les lignes où le titre est vide
    # Note : Vérifiez si votre colonne s'appelle 'Titre' ou 'Job Title'
    col_titre = 'Titre' if 'Titre' in df.columns else 'Job Title'
    df.dropna(subset=[col_titre], inplace=True)

    # Remplacer les textes 'N/A' par de vrais NaN (Not a Number)
    df.replace(['N/A', 'n/a', 'None'], np.nan, inplace=True)

    # Nettoyage des espaces blancs invisibles (\n, \t) dans tout le tableau
    df = df.applymap(lambda x: x.strip() if isinstance(x, str) else x)

    # --- AFFICHAGE ---
    print(f"Nettoyage terminé : {len(df)} lignes conservées.")
    print(df.head())

    # --- SAUVEGARDE ---
    # Sauvegarde locale (plus sûr) ou sur le bureau
    output_path = "jobs_clean.csv"
    df.to_csv(output_path, index=False, encoding='utf-8-sig') # utf-8-sig est idéal pour Excel
    print(f"Fichier sauvegardé avec succès sous : {os.path.abspath(output_path)}")
```
