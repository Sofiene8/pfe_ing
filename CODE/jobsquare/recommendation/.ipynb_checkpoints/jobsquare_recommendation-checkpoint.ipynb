# 🎯 Jobsquare — Système de recommandation Content-Based

## Architecture
```
CV utilisateurs  ──┐
                   ├──► Embeddings ──► Cosine Similarity ──► Top-N offres
Offres d'emploi  ──┘
```

### Étapes du notebook
1. Installation & imports
2. Connexion MongoDB et extraction des données
3. Prétraitement des textes (CV + offres)
4. Génération des embeddings (sentence-transformers)
5. Calcul de la similarité cosinus
6. Moteur de recommandation Top-N
7. Évaluation & visualisation
8. Export vers une API FastAPI (architecture évolutive)

## 3. Prétraitement des textes

````python

# ─────────────────────────────────────────────────────────────────
# Helpers
# ─────────────────────────────────────────────────────────────────

def is_empty(val) -> bool:
    """Retourne True si la valeur est None, NaN, ou chaîne vide."""
    if val is None:
        return True
    if isinstance(val, float) and np.isnan(val):
        return True
    if isinstance(val, str) and not val.strip():
        return True
    return False


def clean_text(text) -> str:
    """Nettoyage basique : HTML, espaces, caractères spéciaux."""
    if is_empty(text):
        return ""
    text = str(text)
    text = re.sub(r"<[^>]+>", " ", text)   # supprimer balises HTML
    text = re.sub(r"[\r\n\t]+", " ", text)  # normaliser sauts de ligne
    text = re.sub(r"\s{2,}", " ", text)     # espaces multiples
    return text.strip()


def safe_str(val, fallback: str = "") -> str:
    """Convertit en str en gérant NaN / None."""
    if is_empty(val):
        return fallback
    return str(val).strip()


# ─────────────────────────────────────────────────────────────────
# Construction du texte utilisateur (CV)
# ─────────────────────────────────────────────────────────────────

def build_user_text(row: pd.Series) -> str:
    """
    Agrège tous les champs disponibles du profil utilisateur
    en un seul texte pour l'embedding.
    """
    parts = []

    # Champs textuels simples
    for field in ["bio", "experience", "education", "resume",
                  "description", "title", "summary"]:
        val = clean_text(row.get(field))
        if val:
            parts.append(val)

    # Compétences (liste ou chaîne)
    skills = row.get("skills", [])
    if isinstance(skills, list):
        skills = [s for s in skills if not is_empty(s)]
        if skills:
            parts.append("Compétences : " + ", ".join(str(s) for s in skills))
    elif isinstance(skills, str) and skills.strip():
        parts.append("Compétences : " + skills.strip())

    # Fallback : au moins le nom ou l'email
    if not parts:
        for fallback_field in ["username", "name", "email"]:
            fb = safe_str(row.get(fallback_field))
            if fb:
                return fb
        return "utilisateur_inconnu"

    return " ".join(parts)


# ─────────────────────────────────────────────────────────────────
# Construction du texte offre d'emploi
# ─────────────────────────────────────────────────────────────────

def build_listing_text(row: pd.Series) -> str:
    """
    Agrège les champs d'une offre d'emploi en un texte pour l'embedding.
    Le titre est répété 2× pour lui donner plus de poids.
    """
    parts = []

    # Titre (pondéré ×2)
    title = clean_text(row.get("title"))
    if title:
        parts.append(f"{title} {title}")

    # Description
    desc = clean_text(row.get("description"))
    if desc:
        parts.append(desc)

    # Compétences requises
    skills = row.get("skills", [])
    if isinstance(skills, list):
        skills = [s for s in skills if not is_empty(s)]
        if skills:
            parts.append("Compétences requises : " + ", ".join(str(s) for s in skills))
    elif isinstance(skills, str) and skills.strip():
        parts.append("Compétences requises : " + skills.strip())

    # Autres champs utiles
    for field in ["category", "location", "contract_type", "sector"]:
        val = clean_text(row.get(field))
        if val:
            parts.append(val)

    return " ".join(parts) if parts else (title or "offre_inconnue")


# ─────────────────────────────────────────────────────────────────
# Application
# ─────────────────────────────────────────────────────────────────

df_users["text"]    = df_users.apply(build_user_text,    axis=1)
df_listings["text"] = df_listings.apply(build_listing_text, axis=1)

# Vérification : aucune valeur NaN ou float ne doit subsister
assert df_users["text"].apply(lambda x: isinstance(x, str)).all(), \
    "df_users contient des valeurs non-str dans la colonne 'text'"
assert df_listings["text"].apply(lambda x: isinstance(x, str)).all(), \
    " df_listings contient des valeurs non-str dans la colonne 'text'"

# ─────────────────────────────────────────────────────────────────
# Aperçu
# ─────────────────────────────────────────────────────────────────

print("✅ Prétraitement terminé\n")
print(f"Utilisateurs : {len(df_users)}  |  Offres : {len(df_listings)}\n")

print("── Exemple texte utilisateur ──────────────────────────────")
print(df_users["text"].iloc[0][:300], "...\n")

print("── Exemple texte offre ────────────────────────────────────")
print(df_listings["text"].iloc[0][:300], "...")

# Diagnostic si un texte est suspect (trop court)
short_users = df_users[df_users["text"].str.len() < 10]
if not short_users.empty:
    print(f"\n⚠️  {len(short_users)} utilisateur(s) avec texte très court :")
    print(short_users[["_id", "text"]].to_string())
````

## 4. Génération des embeddings

````python
# ── Chargement du modèle (téléchargé automatiquement ~120 MB) ─────
print(f"Chargement du modèle : {MODEL_NAME}")
model = SentenceTransformer(MODEL_NAME)

# Alternatives possibles :
#  - 'all-MiniLM-L6-v2'            (anglais uniquement, très rapide)
#  - 'paraphrase-multilingual-mpnet-base-v2' (meilleure qualité, plus lent)
#  - 'distiluse-base-multilingual-cased-v2'  (bon compromis FR/EN)

print(f"✅ Modèle chargé — dimension des vecteurs : {model.get_sentence_embedding_dimension()}")
````

````python
# ── Encodage ──────────────────────────────────────────────────────
print("Encodage des CV utilisateurs...")
user_embeddings = model.encode(
    df_users["text"].tolist(),
    batch_size=32,
    show_progress_bar=True,
    normalize_embeddings=True  # L2-normalisation → dot product = cosine
)

print("\nEncodage des offres d'emploi...")
listing_embeddings = model.encode(
    df_listings["text"].tolist(),
    batch_size=32,
    show_progress_bar=True,
    normalize_embeddings=True
)

print(f"\n✅ Shape embeddings users    : {user_embeddings.shape}")
print(f"✅ Shape embeddings listings : {listing_embeddings.shape}")

# ── Sauvegarde optionnelle pour cache ─────────────────────────────
np.save("user_embeddings.npy",    user_embeddings)
np.save("listing_embeddings.npy", listing_embeddings)
print("\n💾 Embeddings sauvegardés (.npy)")
````

## 5. Matrice de similarité cosinus

````python
print("Colonnes de df_listings :", df_listings.columns.tolist())
print("Colonnes de df_users :", df_users.columns.tolist())
````

````python
# ── Calcul de la matrice (users × listings) ───────────────────────
# Les embeddings sont déjà L2-normalisés → cosine_similarity == dot product
similarity_matrix = cosine_similarity(user_embeddings, listing_embeddings)
LISTING_TITLE_COL = "title"
df_similarity = pd.DataFrame(
    similarity_matrix,
    index=df_users["username"] if "username" in df_users.columns else df_users.index,
    columns=df_listings["title"]
)

print(f"Matrice de similarité : {similarity_matrix.shape}  (users × offres)")
display(df_similarity.round(3))
````

````python
# ── Heatmap de la matrice ─────────────────────────────────────────
fig, ax = plt.subplots(figsize=(max(8, len(df_listings)), max(4, len(df_users))))

sns.heatmap(
    df_similarity,
    annot=True, fmt=".2f",
    cmap="YlOrRd",
    vmin=0, vmax=1,
    linewidths=0.5,
    ax=ax
)
ax.set_title("Matrice de similarité cosinus  (utilisateurs × offres)", pad=14)
ax.set_xlabel("Offres d'emploi")
ax.set_ylabel("Utilisateurs")
plt.xticks(rotation=35, ha="right")
plt.tight_layout()
plt.savefig("similarity_heatmap.png", dpi=150)
plt.show()
print("📊 Heatmap sauvegardée → similarity_heatmap.png")
````

## 6. Moteur de recommandation Top-N

````python
def recommend_for_user(
    user_idx: int,
    similarity_matrix: np.ndarray,
    df_listings: pd.DataFrame,
    top_n: int = 5,
    min_score: float = 0.0
) -> pd.DataFrame:
    """
    Retourne les top_n offres les plus pertinentes pour un utilisateur.

    Paramètres
    ----------
    user_idx         : index de l'utilisateur dans df_users
    similarity_matrix: matrice (n_users, n_listings)
    df_listings      : dataframe des offres
    top_n            : nombre de recommandations
    min_score        : score minimum pour être inclus

    Retourne
    --------
    DataFrame avec colonnes : rank, title, score, _id
    """
    scores   = similarity_matrix[user_idx]              # vecteur (n_listings,)
    top_idx  = np.argsort(scores)[::-1]                 # tri décroissant

    results = []
    rank = 1
    for idx in top_idx:
        if rank > top_n:
            break
        score = float(scores[idx])
        if score < min_score:
            break
        row = df_listings.iloc[idx]
        results.append({
            "rank"    : rank,
            "listing_id": str(row.get("_id", idx)),
            "title"   : row.get("title", f"Offre #{idx}"),
            "location": row.get("location", ""),
            "score"   : round(score, 4),
        })
        rank += 1

    return pd.DataFrame(results)


def recommend_all_users(
    similarity_matrix: np.ndarray,
    df_users: pd.DataFrame,
    df_listings: pd.DataFrame,
    top_n: int = 5
) -> dict:
    """
    Génère les recommandations pour tous les utilisateurs.
    Retourne un dict {user_id: DataFrame_recommendations}
    """
    all_recs = {}
    for i, user_row in df_users.iterrows():
        uid  = str(user_row.get("_id", i))
        name = user_row.get("username", f"user_{i}")
        recs = recommend_for_user(i, similarity_matrix, df_listings, top_n)
        all_recs[uid] = {"username": name, "recommendations": recs}
    return all_recs


print("✅ Fonctions de recommandation définies")
````

````python
# ── Génération des recommandations ────────────────────────────────
all_recommendations = recommend_all_users(
    similarity_matrix, df_users, df_listings, top_n=TOP_N
)
# ── Affichage formaté ─────────────────────────────────────────────
for uid, data in all_recommendations.items():
    print(f"\n{'─'*60}")
    username = data['username'] if isinstance(data['username'], str) and data['username'].strip() else "Inconnu"
    print(f"👤  {username.upper()}  (id: {uid})")
    print(f"{'─'*60}")
    display(data["recommendations"])
````

````python
# ── Requête ponctuelle : nouvel utilisateur ────────────────────────
def recommend_for_new_cv(cv_text: str, df_listings: pd.DataFrame,
                          model, listing_embeddings, top_n=5) -> pd.DataFrame:
    """
    Recommande des offres pour un texte de CV non présent en base.
    Utile pour tester à la volée ou lors de l'inscription.
    """
    vec = model.encode([cv_text], normalize_embeddings=True)
    scores = cosine_similarity(vec, listing_embeddings)[0]
    top_idx = np.argsort(scores)[::-1][:top_n]
    results = []
    for rank, idx in enumerate(top_idx, 1):
        row = df_listings.iloc[idx]
        results.append({
            "rank"    : rank,
            "title"   : row.get("title", f"Offre #{idx}"),
            "location": row.get("location", ""),
            "score"   : round(float(scores[idx]), 4),
        })
    return pd.DataFrame(results)


# ── Exemple d'utilisation ─────────────────────────────────────────
test_cv = """
Ingénieur logiciel avec 5 ans d'expérience en développement web.
Maîtrise de Python, Django, FastAPI, PostgreSQL et Docker.
Expérience en déploiement cloud AWS et CI/CD avec GitHub Actions.
"""

print("Recommandations pour CV test :")
display(recommend_for_new_cv(test_cv, df_listings, model, listing_embeddings, top_n=TOP_N))
````

## 7. Évaluation & visualisation

````python
# ── Distribution des scores de similarité ─────────────────────────
all_scores = similarity_matrix.flatten()
fig, axes = plt.subplots(1, 2, figsize=(12, 4))

# Distribution globale
axes[0].hist(all_scores, bins=30, color="#4A90D9", edgecolor="white", alpha=0.85)
axes[0].set_title("Distribution des scores cosinus")
axes[0].set_xlabel("Score de similarité")
axes[0].set_ylabel("Fréquence")
axes[0].axvline(all_scores.mean(), color="#E55", linestyle="--", label=f"Moyenne = {all_scores.mean():.3f}")
axes[0].legend()

# Scores top-1 par utilisateur
top1_scores = similarity_matrix.max(axis=1)

# ── Fix : convertir tous les labels en str propre ─────────────────
raw_labels = df_users.get("username", pd.Series(range(len(df_users))))
user_labels = [
    str(v).strip() if pd.notna(v) and str(v).strip() != "" else f"user_{i}"
    for i, v in enumerate(raw_labels)
]

axes[1].barh(user_labels, top1_scores, color="#4CAF87", edgecolor="white")
axes[1].set_title("Score max (meilleure offre) par utilisateur")
axes[1].set_xlabel("Score cosinus")
axes[1].set_xlim(0, 1)

for i, v in enumerate(top1_scores):
    axes[1].text(v + 0.01, i, f"{v:.3f}", va="center", fontsize=9)

plt.tight_layout()
plt.savefig("score_distribution.png", dpi=150)
plt.show()
````

## 8. Architecture évolutive — API FastAPI

Le code ci-dessous est **prêt à déployer** en tant que micro-service.
Sauvegarder dans `api/main.py` puis lancer avec `uvicorn api.main:app --reload`.

````python
fastapi_code = '''
"""
api/main.py  —  Jobsquare Recommendation API
Lancement : uvicorn api.main:app --reload --port 8000
"""

from fastapi import FastAPI, HTTPException
from pydantic import BaseModel
from typing import List, Optional
import numpy as np
from sentence_transformers import SentenceTransformer
from sklearn.metrics.pairwise import cosine_similarity
from pymongo import MongoClient

# ── Config ────────────────────────────────────────────────────────
MONGO_URI  = "mongodb://localhost:27017"
DB_NAME    = "jobsquare"
MODEL_NAME = "paraphrase-multilingual-MiniLM-L12-v2"

# ── Initialisation (au démarrage du serveur) ──────────────────────
app   = FastAPI(title="Jobsquare Recommendation API", version="1.0.0")
model = SentenceTransformer(MODEL_NAME)

client = MongoClient(MONGO_URI)
db     = client[DB_NAME]

# ── Schémas Pydantic ──────────────────────────────────────────────
class RecommendationItem(BaseModel):
    rank        : int
    listing_id  : str
    title       : str
    score       : float
    location    : Optional[str] = None

class RecommendRequest(BaseModel):
    cv_text : str
    top_n   : int = 5

class RecommendResponse(BaseModel):
    recommendations : List[RecommendationItem]


# ── Chargement des embeddings offres (cache au démarrage) ─────────
def load_listings():
    listings = list(db["listings"].find({}, {"title":1, "description":1, "skills":1, "location":1}))
    texts = []
    for l in listings:
        title = l.get("title", "")
        desc  = l.get("description", "")
        skills = ", ".join(l.get("skills", []))
        texts.append(f"{title} {title} {desc} Compétences : {skills}")
    embeddings = model.encode(texts, normalize_embeddings=True)
    return listings, embeddings

listings_cache, listing_embeddings_cache = load_listings()


# ── Endpoints ─────────────────────────────────────────────────────
@app.get("/health")
def health():
    return {"status": "ok", "listings_cached": len(listings_cache)}


@app.post("/recommend", response_model=RecommendResponse)
def recommend(request: RecommendRequest):
    """Recommande des offres à partir d'un texte de CV."""
    if not request.cv_text.strip():
        raise HTTPException(400, detail="cv_text est vide")

    vec    = model.encode([request.cv_text], normalize_embeddings=True)
    scores = cosine_similarity(vec, listing_embeddings_cache)[0]
    top_idx = np.argsort(scores)[::-1][:request.top_n]

    results = []
    for rank, idx in enumerate(top_idx, 1):
        l = listings_cache[idx]
        results.append(RecommendationItem(
            rank       = rank,
            listing_id = str(l["_id"]),
            title      = l.get("title", ""),
            score      = round(float(scores[idx]), 4),
            location   = l.get("location"),
        ))
    return RecommendResponse(recommendations=results)


@app.get("/recommend/user/{user_id}", response_model=RecommendResponse)
def recommend_by_user_id(user_id: str, top_n: int = 5):
    """Recommande des offres à partir de l\'id utilisateur MongoDB."""
    from bson import ObjectId
    try:
        user = db["users"].find_one({"_id": ObjectId(user_id)})
    except Exception:
        user = db["users"].find_one({"_id": user_id})

    if not user:
        raise HTTPException(404, detail="Utilisateur introuvable")

    parts = []
    for field in ["bio", "experience", "education", "resume"]:
        if user.get(field):
            parts.append(str(user[field]))
    skills = user.get("skills", [])
    if isinstance(skills, list):
        parts.append("Compétences : " + ", ".join(skills))
    cv_text = " ".join(parts)

    return recommend(RecommendRequest(cv_text=cv_text, top_n=top_n))


@app.post("/cache/refresh")
def refresh_cache():
    """Recharge les embeddings des offres (appeler après ajout de nouvelles offres)."""
    global listings_cache, listing_embeddings_cache
    listings_cache, listing_embeddings_cache = load_listings()
    return {"refreshed": len(listings_cache)}
'''

import os
os.makedirs("api", exist_ok=True)
with open("api/main.py", "w", encoding="utf-8") as f:
    f.write(fastapi_code.strip())

print("✅ API FastAPI générée → api/main.py")
print()
print("Pour lancer l'API :")
print("  pip install fastapi uvicorn")
print("  uvicorn api.main:app --reload --port 8000")
print()
print("Endpoints disponibles :")
print("  GET  /health")
print("  POST /recommend           { cv_text, top_n }")
print("  GET  /recommend/user/{id} ?top_n=5")
print("  POST /cache/refresh")
````
